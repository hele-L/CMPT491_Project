**1. Data Loading and Filtering** 

```python
import numpy as np
import os
import pandas as pd

# to make this notebook's output stable across runs
np.random.seed(42)
```

Data Loading

```python
df_raw = pd.read_csv('brfss_2019_2025_combined.csv')
print("Data Raw shape:", df_raw.shape)
print("Data Raw columns:", df_raw.columns.tolist())
df_raw
```

Data Type and Exploratory

```python
print("===Data Types===")
print(df_raw.dtypes)
```

Age group counts

```python
# Map BRFSS age-group codes to readable labels
age_labels = {
    1: "18-24",
    2: "25-29",
    3: "30-34",
    4: "35-39",
    5: "40-44",
    6: "45-49",
    7: "50-54",
    8: "55-59",
    9: "60-64",
    10: "65-69",
    11: "70-74",
    12: "75-79",
    13: "80+",
    14: "Don't know/Refused",
}

# Count rows in each age group, including missing values as separate category
age_counts = (
    df_raw["_AGEG5YR"]
    .astype("Int64")
    .value_counts(dropna=False)
    .reindex(range(1, 15), fill_value=0)
    .rename(index=age_labels)
)

print("Age group counts for _AGEG5YR:")
print(age_counts)

print("Hello")
```

```python
print("\n===Missing Values for the entire dataset===")
print(df_raw.isnull().sum())

# Overall missing-value percentage in the 18-29 age group
overall_null_pct = df_raw.isnull().mean().mean() * 100
print(f"Overall missing-value rate in the entire dataset: {overall_null_pct:.2f}%")

# Per-column missing-value percentage
null_pct_by_col = (df_raw.isnull().mean() * 100).sort_values(ascending=False)
print("\nMissing-value percentage by column for the entire dataset:")
print(null_pct_by_col)
```

Filter to age group 18 to 29

```python
df_18_29 = df_raw[df_raw["_AGEG5YR"].isin([1, 2])].copy()
df_18_29
```

Filter to age group 55-64

```python
df_55_64 = df_raw[df_raw["_AGEG5YR"].isin([8, 9])].copy()
df_55_64
```

Count Missing Values

```python
print("\n===Missing Values for 18-29 Age Group===")
print(df_18_29.isnull().sum())
```

```python

# Overall missing-value percentage in the 18-29 age group
overall_null_pct_18_29 = df_18_29.isnull().mean().mean() * 100
print(f"Overall missing-value rate in age group 18-29: {overall_null_pct_18_29:.2f}%")

# Per-column missing-value percentage
null_pct_by_col_18_29 = (df_18_29.isnull().mean() * 100).sort_values(ascending=False)
print("\nMissing-value percentage by column for age group 18-29:")
print(null_pct_by_col_18_29)
```

```python
print("\n===Missing Values for 55-64 Age Group===")
print(df_55_64.isnull().sum())
```

```python
# Overall missing-value percentage in the 55-64 age group
overall_null_pct_55_64 = df_55_64.isnull().mean().mean() * 100
print(f"Overall missing-value rate in age group 55-64: {overall_null_pct_55_64:.2f}%")

# Per-column missing-value percentage
null_pct_by_col_55_64 = (df_55_64.isnull().mean() * 100).sort_values(ascending=False)
print("\nMissing-value percentage by column for age group 55-64:")
print(null_pct_by_col_55_64)
```
